# Approximation von Funktionen

```julia
using CairoMakie
using LinearAlgebra

using MMJMesh.Plots
using MMJMesh.Mathematics
```

## Kombination von Funktionen

Taylorpolynom

```julia
x = parameter(0 .. π)
f = sin(x)

n = 9
x0 = π / 6
c = [derivativeat(f, x0, k) / factorial(k) for k = 0:n]
b = [(x - x0)^k for k = 0:n]
Tn = c ⋅ b

fplot(f, Tn)
```

## Stückweise lineare Funktionen - Hutfunktionen

```julia
φs = hatfunctions([0, 1, 3, 4, 5])
fplot(φs[1], φs[3])
```

```julia
c = [1, 0.15, 0.75, 0.3, 0.9]
f = c ⋅ φs
fplot(f)
```

## Lineare Hülle
