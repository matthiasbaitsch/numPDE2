## Schritt 1: Installieren

```julia
using CairoMakie
using MMJMesh.Plots
using MMJMesh.Mathematics

lc(uhat, phis) = sum(uhat .* phis)
```

## Schritt 2: Funktionen approximieren

### Funktionen

```julia
x = parameter(0 .. 4π)

f = cos(x)
g = 1 - x / (4π)
h = f * g

fplot(f, g, h)
```

```julia
fplot(f', g', h')
```

### Funktionswerte

```julia
x = [1, 3, 7]
y = h.(x)
p = fplot(h)
scatter!(x, y, color=:tomato)
p
```

### Basisfunktionen

```julia
nodes = range(0, 4π, 5)
phis = hatfunctions(nodes)
fplot(phis)
```

```julia
fplot(phis[1], phis[4])
```

### Linearkombination

```julia
uHat = [0, 0.5, 0.1, 0.5, 0]
uu = lc(uHat, phis)

fplot(uu)
```

### Approximation

```julia
uHat = h.(nodes)
ha = lc(uHat, phis)
fplot(ha, h)
```

```julia
e = h - ha
fplot(e)
```

### Norm

```julia
norm2(u) = sqrt(integrate(u * u, domain(u)))

norm2(e)
```

### Konvergenz

```julia
function error(Ne)
    nodes = range(0, 4π, Ne + 1)
    phis = hatfunctions(nodes)
    ha = sum(h.(nodes) .* phis)
    return norm2(h - ha)
end

error(4)
```

```julia
Nes = 2 .^ (2:10)
hs = 4π ./ Nes
es = error.(Nes)
```

```julia
scatterlines(Nes, es)
```

```julia
scatterlines(hs, es)
```

```julia
f = Figure()
Axis(f[1, 1], xscale=log2, yscale=log10)
scatterlines!(Nes, es)
f
```

```julia
f = Figure()
Axis(f[1, 1], xscale=log10, yscale=log10)
scatterlines!(hs, es)
f
```
