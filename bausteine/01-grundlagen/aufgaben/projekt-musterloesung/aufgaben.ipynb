## Balken auf Federn

### Lösungskandidat und Testfunktion

```julia
l = 3.7
EI = 2.2
k = 3.9
q = 1.1

x = parameter(0 .. l)
w = q * (12 * EI * l + k * l^3 * x - 2 * k * l * x^3 + k * x^4) / (24 * EI * k)
δw = sin(x) + x

fplot(w, δw, zero(x))
```

### Schwache Formulierung

Linear und Bilinearform

```julia
a(w, δw) = EI * integrate(w'' * δw'', 0 .. l) + k * w(0) * δw(0) + k * w(l) * δw(l)
b(δw) = q * integrate(δw, 0 .. l)
```

Kontrolle mit Testfunktion

```julia
println(a(w, δw) ≈ b(δw))
```

### Kontrolle Randwertproblem

Differentialgleichung

```julia
fplot(EI * w'''' - q)
```

Randbedingungen

```julia
println(k * w(0) + EI * w'''(0))
println(k * w(l) - EI * w'''(l))
println(w''(0))
println(w''(l))
```
