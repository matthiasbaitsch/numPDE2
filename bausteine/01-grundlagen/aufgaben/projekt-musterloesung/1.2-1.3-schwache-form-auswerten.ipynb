## Aufgabe 2

```julia
l = 10
EA = 1
C = 2
S = 5
n = 1
F = 10

x = parameter(0 .. l)
u = 10 - x / 20
δu = x

fplot(u, δu)
```

```julia
a(u, δu) = EA * integrate(u' * δu', 0 .. l) + C * integrate(u * δu, 0 .. l) + S * u(l) * δu(l)
b(δu) = n * integrate(δu, 0 .. l) + F * δu(0)

println(a(u, δu))
println(b(δu))
```

```julia
a(u, u)
```

### Aufgabe 3

```julia
l = 3.7
EI = 2.2
k = 3.9
q = 1.1

x = parameter(0 .. l)
w = (12 * EI * l * q + k * l^3 * q * x - 2 * k * l * q * x^3 + k * q * x^4) / (24 * EI * k)
δw = 1 + x

fplot(w, δw, zero(x))
```

```julia
a(w, δw) = EI * integrate(w'' * δw'', 0 .. l) + k * w(0) * δw(0) + k * w(l) * δw(l)
b(δw) = q * integrate(δw, 0 .. l)

δw = sin(x) + x
println(a(w, δw))
println(b(δw))
```

```julia
fplot(EI * w'''' - q)
```

```julia
println(k * w(0) + EI * w'''(0))
println(k * w(l) - EI * w'''(l))
println(w''(0))
println(w''(l))
```
