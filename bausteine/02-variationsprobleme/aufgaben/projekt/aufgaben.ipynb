## Funktionale, Linear- und Bilinearformen

### Funktionen zum Testen

```julia
l = 1.5                   # Irgendeine Länge
x = parameter(0 .. l)     # Parameter
u1 = sin(3*x)             # Sinusfunktion
u2 = x                    # Lineare Funktion
u3 = constant(x, 1.0)     # Konstante Funktion
u4 = 0 * x                # Nullfunktion

fplot(u1, u2, u3)
```

Ein Beispiel

```julia
a(u, v) = integrate(u' * v', 0 .. l)

println("l1: ", a(u1 + u2, u3) ≈ a(u1, u3) + a(u2, u3))
println("l2: ", a(u1, u2 + u3) ≈ a(u1, u2) + a(u1, u3))
println("l3: ", a(3 * u1, 4*u2) ≈ 12*a(u1, u2))
println(" s: ", a(u1, u2) ≈ a(u2, u1))
println("pd: ", a(u3, u3) > 0)
```
